# Part 0 : Q-Learning with Tables






# Q-Table Learning

## Introduction

Ce notebook contient un code Python qui implémente l'algorithme de Q-learning
tabulaire pour entraîner un agent à naviguer dans l'environnement "FrozenLake-v1" de OpenAI Gym. L'objectif de l'agent est d'apprendre à atteindre la case "G" (but) en partant de la case "S" (départ), en évitant de tomber dans les cases "H" (trous) sur un lac gelé représenté par des cases "F" (surface gelée sûre). L'agent apprendra une stratégie optimale pour maximiser ses chances d'atteindre le but.

```python
import gym # Importation de la bibliothèque Gym (OpenAI Gym) qui fournit des environnements pour l'apprentissage par renforcement
import numpy as np # Importation la bibliothèque Numpy et la renomme 'np'.Elle est utile pour les calculs numériques
```


### Load the environment

```python
env = gym.make('FrozenLake-v1', new_step_api=True) #Création de l'environnement "FrozenLake-v1" de Gym, qui simule le jeu du lac gelé.
```

## Concepts Clés

Pour comprendre ce code, il est important de connaître quelques concepts de base de l'apprentissage par renforcement :

**Agent :** C'est l'entité qui apprend à prendre des décisions.

**Environnement :** C'est le monde virtuel dans lequel l'agent interagit et apprend (ici, le jeu "FrozenLake-v1").

**État (State) :** C'est une représentation de la situation actuelle de l'agent dans l'environnement (ici, la case où se trouve l'agent dans FrozenLake).

**Action :** C'est une action que l'agent peut réaliser dans l'environnement (ici, se déplacer vers le haut, le bas, la gauche ou la droite dans FrozenLake).

**Récompense (Reward) :** C'est un signal que l'environnement envoie à l'agent pour indiquer si son action était "bonne" ou "mauvaise" (ici, récompense de 1 quand l'agent atteint le but "G", 0 sinon).

**Q-Learning :** C'est un algorithme d'apprentissage par renforcement qui permet à l'agent d'apprendre une "Q-table" qui estime la "qualité" (valeur Q) de chaque action dans chaque état.

**Q-Table :** C'est un tableau (matrice) qui stocke les valeurs Q. Chaque ligne correspond à un état, et chaque colonne à une action. La valeur Q Q[état, action] représente l'estimation de la récompense future cumulée si l'agent choisit l'action action dans l'état état.

### Implement Q-Table learning algorithm

```python
Q = np.zeros([env.observation_space.n, env.action_space.n]) # On initialise la table Q avec des zéros
# Gestion des paramètres d'apprentissage
lr = 0.8  # lr (learning rate) est le taux d'apprentissage. Il contrôle à quelle vitesse notre agent apprend de nouvelles informations.
y = 0.95  # y (gamma) est le facteur d'escompte.  Plus gamma est élevé, plus l'agent se soucie des récompenses futures (comme atteindre le but à la fin du jeu).
num_episodes = 2000 # num_episodes est le nombre d'épisodes d'entraînement. Un épisode représente une partie complète du jeu. On va entraîner l'agent pendant 2000 parties.


rList = [] # liste pour enregistrer les récompenses totales obtenues à chaque partie de jeu.

for i in range(num_episodes): # Boucle principale qui s'exécute pour chaque partie de jeu, elle prend en compte le nombre total de parties.
    s = env.reset() # Réinitialisation de l'environnement FrozenLake au début de chaque épisode
    rAll = 0 # Initialisation de la variable 'rAll' (récompense totale obtenu lors de la partie  à zéro au début de chaque partie).
    done = False # Initialisation de la variable 'done' à False. 'done' deviendra True quand la partie  est terminé (agent atteint le but ou tombe dans un trou).
    j = 0 # Initialisation de la variable 'j'.Sert à limiter la longueur de chaque partie.
    while j < 99: # boucle while qui va continuer à s'exécuter tant que le nombre de pas de temps j est inférieur à 99
        j += 1    # On incrémente le compteur de pas de temps 'j' de 1 à chaque itération de la boucle 'while
        a = np.argmax(Q[s, :] + np.random.randn(1, env.action_space.n) * (1. / (i + 1))) # Choix d'une action gloutonne (avec bruit) à partir de la table Q
        s1, r, terminated, truncated, _ = env.step(a) # L'agent fait une action dans le jeu et regarde ce qui se passe.
        done = terminated or truncated # On regarde si le jeu est fini ou pas
        Q[s, a] = Q[s, a] + lr * (r + y * np.max(Q[s1, :]) - Q[s, a]) # Mise à jour de la table Q-table pour apprendre du jeu
        rAll += r # Ajout de la récompense de ce pas à la récompense totale de l'épisode.
        s = s1 # L'état actuel devient le nouvel état pour le prochain pas.
        if done: # Si le jeu est fini.
            break # alors on arrête la partie de jeu maintenant.
    rList.append(rAll) # Ajout de la récompense totale de cette partie à la liste des récompenses.
```

```python
print ("Score over time: " +  str(sum(rList)/num_episodes)) # Affiche le score moyen de l'agent sur toutes les parties.
```

```python
print ("Final Q-Table Values")
print (Q) # On affiche la Q-table (le tableau qui contient ce que l'agent a appris).
```

## Analyse des résultats

###  Interprétation du Score Moyen

Le score moyen obtenu par l'agent après 2000 parties est de 0.4065.Ce score de 0.4065 signifie qu'en moyenne, sur chaque partie de jeu, l'agent a accumulé une récompense totale de 0.4065.Dans le contexte du jeu FrozenLake, où la récompense pour atteindre le but est de 1 et 0 sinon, un score de 0.4065 signifie que l'agent réussit à atteindre le but dans environ 40% des parties, en moyenne.

### Évaluation de la Performance de l'Agent

À mon avis, la performance de l'agent, avec un score moyen de 0.4065, peut être considérée comme faible.En effet, un agent parfait, qui réussirait à atteindre le but à chaque partie, aurait un score moyen proche de 1. Un score de 0.4065 est donc encore assez éloigné de cette performance idéale.Cependant, il est important de prendre en compte qu'au début de l'entraînement, l'agent ne connaissait rien du jeu et agissait de manière aléatoire. Le score de 0.4065 représente donc déjà un certain niveau d'apprentissage et une amélioration par rapport au début ou il ne connaissait rien du tout.Globalement, la performance de l'agent est en courageante,cela montre que le score peut-être encore largement améliorer.
